# Reproduce the cold-chain warehouse-location study
Run the cells in order. In Colab, select the delivered Cold_Chain_Warehouse_Location.zip when requested. Locally, open the notebook from its notebooks folder. The default run reproduces the approved manuscript.

```python
from pathlib import Path
import os, zipfile
candidates = [Path.cwd(), Path.cwd().parent, Path('/content/Cold_Chain_Warehouse_Location')]
root = next((p for p in candidates if (p/'run_analysis.py').is_file()), None)
if root is None:
    from google.colab import files
    uploaded = files.upload()
    archive = next(name for name in uploaded if name.endswith('.zip'))
    with zipfile.ZipFile(archive) as z:
        z.extractall('/content/cold_chain_package')
    root = next(Path('/content/cold_chain_package').rglob('run_analysis.py')).parent
os.chdir(root)
print(root)
```
